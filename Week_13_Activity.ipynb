```python
#Task_01_Matplotlib

import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_excel('applications.xlsx')
print(df.head())
print(df.isnull().sum())
```

```python
apps_by_college = df.groupby('College')['Applicant'].count()

plt.bar(apps_by_college.index, apps_by_college.values,
        color='skyblue')

plt.title('Applicants by College')
plt.xlabel('College')
plt.ylabel('Applicants')

plt.show()
print('Most Applicants:',apps_by_college.idxmax(),apps_by_college.max())
```

```python
#Task_01_Seaborn

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

sns.barplot(data=df, x='College', y='Applicant', color='pink')     #Couldnt quite get this to work.
 
plt.title('Applications by College')
plt.xlabel('College')
plt.ylabel('Applicant')

plt.show()
```

```python
#Task_01_Plotly

import plotly.express as px

apps_by_college2 = df.groupby('College',as_index=False)['Applicant'].count()

fig = px.bar(apps_by_college2, x='College', y='Applicant',
             color_discrete_sequence=['purple'],
             title='Applicants by College')

fig.update_layout(
    xaxis_title='College',
    yaxis_title='Applicant'
)

fig.show()
```

```python
#Task_02_Matplotlib

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))

frequency, intervals, bars = plt.hist(df['SAT_ACT'], bins=6, color='pink')

plt.title('SAT_ACT Distribution')
plt.xlabel('SAT_ACT')
plt.ylabel('Frequency')

plt.show()
print("Highest Frequency:", frequency.max())
```

```python
#Task_02_Seaborn

import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np

plt.figure(figsize=(8, 5))
sns.histplot(data=df, x='SAT_ACT', bins=6,
             color='skyblue', edgecolor='black')

plt.title('Distribution SAT_ACT')
plt.xlabel('SAT_ACT')
plt.ylabel('Frequency')

plt.show()
```

```python
#Task_03_Plotly

import plotly.express as px
import numpy as np

fig = px.histogram(
    df,
    x='SAT_ACT',
    nbins=6,
    color_discrete_sequence=['blue'],
    title='Distribution of SAT_ACT',
    width=800,
    height=500
)

fig.update_traces(xbins=dict(start=400, end=1600, size=200))

fig.update_layout(
    xaxis_title='SAT_ACT',
    yaxis_title='Frequency'
)

fig.show()
```

```python
#Task_03_Matplotlib

import matplotlib.pyplot as plt
import numpy as np

plt.figure(figsize=(8, 5))
plt.scatter(df['SAT_ACT'], df['Extracurricular_Hours'], color='blue')

m, b = np.polyfit(df['SAT_ACT'], df['Extracurricular_Hours'], 1)
plt.plot(df['SAT_ACT'], m * df['SAT_ACT'] + b, color='red')

plt.title('SAT_ACT vs Extracurricular_Hours')
plt.xlabel('SAT_ACT')
plt.ylabel('Extracurricular_Hours')

plt.show()     #This is a bit of a mess, I am not sure if my code is wrong or if this is what it is suppose to look like.
```

```python
#Task_03_Seaborn

import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))
sns.regplot(data=df, x='SAT_ACT', y='Extracurricular_Hours',
            scatter_kws={'color': 'blue'},
            line_kws={'color': 'red'},
            ci=None)

plt.title('SAT_ACT vs Extracurricular_Hours')
plt.xlabel('SAT_ACT')
plt.ylabel('Extracurricular_Hours')

plt.show()
```

```python
#Task_03_Plotly

import plotly.express as px

fig = px.scatter(
    df,
    x='SAT_ACT',
    y='Extracurricular_Hours',
    trendline='ols',
    color_discrete_sequence=['blue'],
    title='SAT_ACT vs Extracurricular_Hours',
    width=800,
    height=500
)

fig.update_layout(
    xaxis_title='SAT_ACT',
    yaxis_title='Extracurricular_Hours'
)

fig.show()
```
